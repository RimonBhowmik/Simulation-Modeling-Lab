```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

Part 1

```python
np.random.seed(42)
regular_day_wait_times = np.random.uniform(low=3, high=15, size=1000)

print("Generated Uniform Wait Times (first 10):", regular_day_wait_times[:10])

mean_val = np.mean(regular_day_wait_times)
median_val = np.median(regular_day_wait_times)
std_dev_val = np.std(regular_day_wait_times)

print(f"Mean: {mean_val:.2f} minutes")
print(f"Median: {median_val:.2f} minutes")
print(f"Standard Deviation: {std_dev_val:.2f} minutes")
```

Part 2

```python
wait_times_raw = np.random.normal(loc=9, scale=2, size=1000)

wait_times = np.maximum(0, wait_times_raw)

mean_val = np.mean(wait_times)
median_val = np.median(wait_times)
std_dev_val = np.std(wait_times)
max_val = np.max(wait_times)

print(f"Mean: {mean_val:.2f} minutes")
print(f"Median: {median_val:.2f} minutes")
print(f"Standard Deviation: {std_dev_val:.2f} minutes")
print(f"Maximum: {max_val:.2f} minutes")
```

Part C

```python
# Plot histograms for both distributions in the same figure (two subplots).

plt.figure(figsize=(12, 5))

# Subplot 1: Regular Day Wait Times (Uniform Distribution)
plt.subplot(1, 2, 1)
plt.hist(regular_day_wait_times, bins=30, edgecolor='black', alpha=0.7, color='blue')
plt.title('Regular Day Wait Times (Uniform Distribution)*-*')
plt.xlabel('Wait Time (minutes)')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)

# Subplot 2: Busy Day Wait Times (Normal Distribution)
plt.subplot(1, 2, 2)
plt.hist(wait_times, bins=30, edgecolor='black', alpha=0.7, color='Red')
plt.title('Busy Day Wait Times (Normal Distribution)*-*')
plt.xlabel('Wait Time (minutes)')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)

plt.tight_layout()
plt.show()
```
